```python
import numpy as np
import matplotlib.pyplot as plt

from gpa import GPA
```

```python
# Grid 9 × 9
x = np.arange(9)
y = np.arange(9)

X, Y = np.meshgrid(x, y)

# Center of the pattern
cx, cy = 4, 4

# Gaussian profile
sigma = 1.8

ex = np.exp(
    -((X - cx)**2 + (Y - cy)**2) / (2 * sigma**2)
)

# Add a small asymmetric component
# ex += 1 * np.exp(
#     -((X - 6)**2 + (Y - 4)**2) / (2 * 0.8**2)
# )
```

```python
ny, nx = ex.shape

fig, ax = plt.subplots(figsize=(7, 6))

# Image
ax.imshow(
    ex,
    origin="lower",
    cmap="gray"
)

# Show center
ax.scatter(
    cx ,
    cy ,
    marker="x",
    s=100,
    linewidths=2
)

plt.tight_layout()
plt.show()
```

```python
gpa = GPA(ex)
result = gpa.evaluate()

print(result)
```
